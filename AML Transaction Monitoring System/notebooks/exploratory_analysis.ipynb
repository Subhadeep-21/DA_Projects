```python
# ============================================
# AML TRANSACTION MONITORING - EXPLORATORY DATA ANALYSIS
# ============================================

# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
# Display settings
pd.set_option('display.max_columns', None)

# ============================================
# LOAD DATA
# ============================================

# Load JSON dataset
df = pd.read_json('csvjson (1).json')

# Preview dataset
print("First 5 Rows:")
display(df.head())

# Dataset Shape
print(f"\nDataset Shape: {df.shape}")

# ============================================
# DATASET INFORMATION
# ============================================

print("\nDataset Information:")
df.info()

# ============================================
# CHECK MISSING VALUES
# ============================================

print("\nMissing Values:")
missing_values = df.isnull().sum()
display(missing_values)

# ============================================
# CHECK DUPLICATES
# ============================================

duplicates = df.duplicated().sum()

print(f"\nDuplicate Records: {duplicates}")

# ============================================
# DATA TYPE CONVERSION
# ============================================

# Convert TransactionDate to datetime
df['TransactionDate'] = pd.to_datetime(
    df['TransactionDate'],
    format='mixed',
    errors='coerce'
)

# Create additional date columns
df['TransactionMonth'] = df['TransactionDate'].dt.month_name()
df['TransactionHour'] = df['TransactionDate'].dt.hour
df['TransactionDay'] = df['TransactionDate'].dt.day_name()

# ============================================
# DESCRIPTIVE STATISTICS
# ============================================

print("\nDescriptive Statistics:")
display(df.describe())

# ============================================
# UNIQUE VALUES
# ============================================

print("\nUnique Transaction Types:")
print(df['TransactionType'].unique())

print("\nUnique Channels:")
print(df['Channel'].unique())

print("\nUnique Occupations:")
print(df['CustomerOccupation'].unique())

# ============================================
# TOTAL TRANSACTION VOLUME
# ============================================

total_volume = df['TransactionAmount'].sum()

print(f"\nTotal Transaction Volume: ${total_volume:,.2f}")

# ============================================
# TRANSACTION TYPE ANALYSIS
# ============================================

transaction_type_count = df['TransactionType'].value_counts()

print("\nTransaction Type Distribution:")
display(transaction_type_count)

# Plot Transaction Type Distribution
plt.figure(figsize=(6,4))
transaction_type_count.plot(kind='bar')
plt.title('Transaction Type Distribution')
plt.xlabel('Transaction Type')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()

# ============================================
# CHANNEL ANALYSIS
# ============================================

channel_analysis = df['Channel'].value_counts()

print("\nChannel Distribution:")
display(channel_analysis)

# Plot Channel Distribution
plt.figure(figsize=(6,4))
channel_analysis.plot(kind='bar')
plt.title('Channel Distribution')
plt.xlabel('Channel')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()

# ============================================
# TOP LOCATIONS BY TRANSACTIONS
# ============================================

top_locations = df['Location'].value_counts().head(10)

print("\nTop 10 Locations:")
display(top_locations)

# Plot Top Locations
plt.figure(figsize=(10,5))
top_locations.plot(kind='bar')
plt.title('Top 10 Transaction Locations')
plt.xlabel('Location')
plt.ylabel('Transaction Count')
plt.xticks(rotation=45)
plt.show()

# ============================================
# HIGH VALUE TRANSACTIONS
# ============================================

high_value_txns = df[df['TransactionAmount'] > 1000]

print(f"\nHigh Value Transactions (>1000): {len(high_value_txns)}")

display(high_value_txns.head())

# ============================================
# LOGIN ATTEMPT ANALYSIS
# ============================================

login_attempts = df['LoginAttempts'].value_counts().sort_index()

print("\nLogin Attempts Distribution:")
display(login_attempts)

# Plot Login Attempts
plt.figure(figsize=(6,4))
login_attempts.plot(kind='bar')
plt.title('Login Attempts Distribution')
plt.xlabel('Login Attempts')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()

# ============================================
# SUSPICIOUS TRANSACTION DETECTION
# ============================================

# Create Suspicious Flag
df['SuspiciousFlag'] = np.where(
    (df['TransactionAmount'] > 1000) |
    (df['LoginAttempts'] >= 3) |
    (df['AccountBalance'] < df['TransactionAmount']) |
    (df['TransactionDuration'] < 20),
    1,
    0
)

# Count suspicious transactions
suspicious_count = df['SuspiciousFlag'].sum()

print(f"\nSuspicious Transactions Detected: {suspicious_count}")

# View suspicious transactions
suspicious_txns = df[df['SuspiciousFlag'] == 1]

display(suspicious_txns.head())

# ============================================
# CHANNEL-WISE SUSPICIOUS ACTIVITY
# ============================================

channel_risk = suspicious_txns['Channel'].value_counts()

print("\nSuspicious Transactions by Channel:")
display(channel_risk)

# Plot Channel Risk
plt.figure(figsize=(6,4))
channel_risk.plot(kind='bar')
plt.title('Suspicious Transactions by Channel')
plt.xlabel('Channel')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()

# ============================================
# OCCUPATION-WISE RISK ANALYSIS
# ============================================

occupation_risk = suspicious_txns['CustomerOccupation'].value_counts()

print("\nOccupation-wise Suspicious Activity:")
display(occupation_risk)

# Plot Occupation Risk
plt.figure(figsize=(8,4))
occupation_risk.plot(kind='bar')
plt.title('Occupation-wise Suspicious Transactions')
plt.xlabel('Occupation')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()

# ============================================
# DEVICE REUSE ANALYSIS
# ============================================

device_reuse = df.groupby('DeviceID')['AccountID'].nunique()

shared_devices = device_reuse[device_reuse > 1]

print(f"\nDevices Used by Multiple Accounts: {len(shared_devices)}")

display(shared_devices.head())

# ============================================
# IP ADDRESS REUSE ANALYSIS
# ============================================

ip_reuse = df.groupby('IP Address')['AccountID'].nunique()

shared_ips = ip_reuse[ip_reuse > 1]

print(f"\nIP Addresses Used by Multiple Accounts: {len(shared_ips)}")

display(shared_ips.head())

# ============================================
# MONTHLY TRANSACTION TREND
# ============================================

monthly_trend = df.groupby('TransactionMonth')['TransactionAmount'].sum()

print("\nMonthly Transaction Trend:")
display(monthly_trend)

# Plot Monthly Trend
plt.figure(figsize=(12,5))
monthly_trend.plot(kind='line', marker='o')
plt.title('Monthly Transaction Volume Trend')
plt.xlabel('Month')
plt.ylabel('Transaction Volume')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()

# ============================================
# TRANSACTION AMOUNT DISTRIBUTION
# ============================================

plt.figure(figsize=(8,5))
plt.hist(df['TransactionAmount'], bins=30)
plt.title('Transaction Amount Distribution')
plt.xlabel('Transaction Amount')
plt.ylabel('Frequency')
plt.show()

# ============================================
# CUSTOMER AGE ANALYSIS
# ============================================

plt.figure(figsize=(8,5))
plt.hist(df['CustomerAge'], bins=20)
plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()

# ============================================
# RISK SCORE ENGINE
# ============================================

def calculate_risk_score(row):
    
    score = 0
    
    if row['TransactionAmount'] > 1000:
        score += 40
        
    if row['LoginAttempts'] >= 3:
        score += 30
        
    if row['Channel'] == 'Online':
        score += 10
        
    if row['TransactionDuration'] < 20:
        score += 20
        
    if row['AccountBalance'] < row['TransactionAmount']:
        score += 30
        
    return score

# Apply Risk Score
df['RiskScore'] = df.apply(calculate_risk_score, axis=1)

# Risk Category
df['RiskCategory'] = pd.cut(
    df['RiskScore'],
    bins=[-1, 30, 60, 100],
    labels=['Low', 'Medium', 'High']
)

print("\nRisk Category Distribution:")
display(df['RiskCategory'].value_counts())

# ============================================
# TOP HIGH RISK TRANSACTIONS
# ============================================

high_risk_txns = df.sort_values(by='RiskScore', ascending=False)

print("\nTop High Risk Transactions:")
display(
    high_risk_txns[
        [
            'TransactionID',
            'AccountID',
            'TransactionAmount',
            'Channel',
            'LoginAttempts',
            'RiskScore',
            'RiskCategory'
        ]
    ].head(10)
)

# ============================================
# EXPORT CLEANED DATA
# ============================================

# Save processed data
df.to_csv('aml_processed_data.csv', index=False)

print("\nProcessed AML dataset exported successfully!")

# ============================================
# KEY INSIGHTS
# ============================================

print("""
============================================
KEY AML INSIGHTS
============================================

1. Online and ATM channels show elevated risk activity.

2. Multiple login attempts may indicate account takeover attempts.

3. Shared IP addresses and Device IDs could signal mule/fraud networks.

4. High transaction amounts combined with low balances are potential AML red flags.

5. Risk scoring helps prioritize suspicious transactions for investigation.

============================================
EDA COMPLETED SUCCESSFULLY
============================================
""")
```
